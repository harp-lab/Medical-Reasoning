# Scratch: simplest Claude API call

Loads `APIKEY` from `.env` and sends one message. Requires a positive credit balance
on the org that owns the key (otherwise you'll see `invalid_request_error: credit balance too low`).

```python
import os
import anthropic
from dotenv import load_dotenv

load_dotenv()                       # reads .env -> APIKEY
client = anthropic.Anthropic(api_key=os.environ["APIKEY"])

msg = client.messages.create(
    model="claude-sonnet-4-5",
    max_tokens=200,
    messages=[
        {"role": "user", "content": "In one sentence, what is Parkinson's disease?"}
    ],
)

print(msg.content[0].text)
```
